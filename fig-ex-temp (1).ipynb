## 抛物线的优化问题

补充案例

步长 $\eta = 0.25$，初始值 $x_0 = 3$，迭代 7 步收敛；图 2-3 (b) 中，$\eta = 0.25$，$x_0 = -2$，迭代 8 步收敛；图 2-3 (c) 中，$\eta = 0.1$，$x_0 = 0$，迭代 15 步收敛；图 2-3 (d) 中，$\eta = 0.5$，$x_0 = 4$，迭代 2 步收敛。尽管迭代步数会受步长和初始值影响，但是最终都收敛到抛物线的谷底，即达到了函数最小值。


$f(x)=\min (x-1)^2 $ 

https://www.yuque.com/sdutzhou/durg6g/xlgwtny2k6golq9h?singleDoc# 《迭代优化原理》

```python
import numpy as np
import matplotlib.pyplot as plt

def grad(X):                  # 定义梯度
    return 2 * X - 2

def gradient_descent(X0, ita):    # 定义梯度下降函数
    Val = [X0]  # 初始值直接存入列表
    f = X0 **2 - 2 * X0 + 1
    Func = [f]

    count = 0
    iter_max = 100  # 限制最大迭代次数，避免无限循环
    epsilon = 1e-3  # 收敛阈值
    X = X0

    while count < iter_max:
        # 计算下一步
        up_X = X - ita * grad(X)
        up_f = up_X** 2 - 2 * up_X + 1
        # 计算函数值变化
        delta = abs(f - up_f)
        # 更新变量
        X, f = up_X, up_f
        Val.append(X)
        Func.append(f)
        count += 1
        # 满足收敛条件则停止
        if delta < epsilon:
            break

    return Val, Func, count

if __name__ == "__main__":
    x_min, x_max = -3, 5  # 适当限制x轴范围，避免图像过大
    sample_num = 100
    dot_x = np.linspace(x_min, x_max, sample_num)
    dot_y = dot_x ** 2 - 2 * dot_x + 1  # 函数：f(x)=(x-1)^2

    X0 = -2    # 初始值
    ita = 0.25 # 步长

    Val, Func, count = gradient_descent(X0, ita)

    # 打印迭代信息
    np.set_printoptions(formatter={'float': '{: 0.3f}'.format})
    print("迭代点坐标：", np.array(Val))
    print("迭代点函数值：", np.array(Func))
    print(f"步长=%.3f, 初始值=%.1f, 总迭代次数=%d" % (ita, X0, count))

    # 绘图
    plt.figure(figsize=(8, 6))  # 固定图像尺寸，避免自动调整过大
    plt.plot(dot_x, dot_y, "-", c='b', label=r'$f(x)=(x-1)^2$')
    plt.scatter(Val, Func, color='r', s=40, marker='o', label='迭代点')

    # 标注迭代点
    for i in range(len(Val)):
        # 限制标注位置，避免超出图像范围
        plt.text(Val[i], Func[i] + 0.3, f'$x_{i}$', fontsize=9)

    plt.xlim(x_min, x_max)  # 固定x轴范围
    plt.ylim(min(dot_y) - 1, max(Func) + 1)  # 固定y轴范围
    plt.title(r'$\eta$=%.3f, $x_0$=%.1f, 迭代次数=%d' % (ita, X0, count), fontsize=12)
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()
```

## 函数$y=-ln(x)$ 的梯度消失演示

步长$\eta = 1$，阈值$\varepsilon = 0.1$，迭代两步收敛；图 2-4 (b) 中，$\eta = 1$，$\varepsilon = 0.01$，迭代 38 步收敛；图 2-4 (c) 中，$\eta = 0.1$，$\varepsilon = 0.1$，迭代 5 步收敛；图 2-4 (d) 中，$\eta = 0.1$，$\varepsilon = 0.01$，迭代 49 步收敛。由此可见，当步长固定时，阈值对收敛前迭代次数起决定性作用；当阈值固定时，步长的变化对迭代次数影响不大，但是在曲线陡峭的部分，即$x \in (0, 6)$范围内，步长越大，从$x_0$到$x_1$之间的跨度也就越大。

https://www.yuque.com/sdutzhou/durg6g/qgvb0ngfo2lu7hnc?singleDoc# 《梯度消失例题》

```python
import numpy as np
import math
import matplotlib.pyplot as plt

def grad(X):                  # 定义梯度（f(x) = -lnx 的导数为 -1/x ）
    return -1 / X  # 注意：f(x) = -lnx 的导数是 -1/x，梯度下降公式中是减梯度，即 X - ita*(-1/x) = X + ita/x？需确认函数梯度是否正确

def gradient_descent(X0, ita):    # 定义梯度下降函数
    Val = [X0]  # 初始值直接存入列表
    f = -1 * math.log(X0)  
    Func = [f]

    count = 0
    iter_max = 100  # 限制最大迭代次数，避免无限循环
    epsilon = 0.1  # 收敛阈值
    X = X0

    while count < iter_max:
        # 计算下一步（注意：若 f(x) = -lnx，梯度是 -1/x，梯度下降应为 X - ita*(-1/x) = X + ita/x）
        # 此处根据原代码的 grad 函数继续，但需注意数学正确性
        up_X = X - ita * grad(X)
        # 确保 x 为正数（ln(x) 定义域为 x>0）
        if up_X <= 0:
            print("迭代点 x 小于等于 0，超出 ln(x) 定义域，停止迭代")
            break
        up_f = -1 * math.log(up_X)
        # 计算函数值变化
        delta = abs(f - up_f)
        # 更新变量
        X, f = up_X, up_f
        Val.append(X)
        Func.append(f)
        count += 1
        # 满足收敛条件则停止
        if delta < epsilon:
            break

    return Val, Func, count

if __name__ == "__main__":
    x_min, x_max = 0.01, 10  # 修正 x_min 为 0.01（避免 ln(0) 无意义）
    sample_num = 200
    dot_x = np.linspace(x_min, x_max, sample_num)
    dot_y = -1 * np.log(dot_x)  # 改用 np.log() 处理数组

    X0 = 0.2   # 初始值（确保为正数）
    ita = 1  # 步长

    Val, Func, count = gradient_descent(X0, ita)

    # 打印迭代信息
    np.set_printoptions(formatter={'float': '{: 0.3f}'.format})
    print("迭代点坐标：", np.array(Val))
    print("迭代点函数值：", np.array(Func))
    print(f"步长=%.3f, 初始值=%.1f, 总迭代次数=%d" % (ita, X0, count))

    # 绘图
    plt.figure(figsize=(8, 6))
    plt.plot(dot_x, dot_y, "-", c='b', label=r'$f(x)=-\ln(x)$')
    plt.scatter(Val, Func, color='r', s=40, marker='o', label='迭代点')

    # 标注迭代点
    for i in range(len(Val)):
        plt.text(Val[i], Func[i] + 0.3, f'$x_{i}$', fontsize=9)

    plt.xlim(x_min, x_max)
    plt.ylim(min(dot_y) - 1, max(Func) + 1)
    plt.title(r'$\eta$=%.3f, $x_0$=%.1f, 迭代次数=%d' % (ita, X0, count), fontsize=12)
    plt.xlabel('x')
    plt.ylabel('f(x)')
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()
```

```python
#2.3节 例4的Python程序   
import matplotlib.pyplot as plt
import numpy as np

# 设置中文显示
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False  # 正确显示负号

def gradient_descent_2d(alpha, x0=3, y0=-1, epsilon=0.01):
    """
    对二元函数f(x,y)=(x-1)²+(y+2)²执行梯度下降
    参数:
        alpha: 学习率
        x0, y0: 初始点坐标
        epsilon: 收敛阈值
    返回:
        迭代次数列表、x值历史、y值历史
    """
    x, y = x0, y0
    iterations = 0
    x_history = [x]
    y_history = [y]
    iter_history = [0]
    
    while True:
        # 计算梯度
        grad_x = 2 * (x - 1)  # x方向梯度
        grad_y = 2 * (y + 2)  # y方向梯度
        
        # 保存当前值
        x_old, y_old = x, y
        
        # 参数更新
        x = x - alpha * grad_x
        y = y - alpha * grad_y
        
        iterations += 1
        x_history.append(x)
        y_history.append(y)
        iter_history.append(iterations)
        
        # 计算变化量（使用欧氏距离）
        delta = np.sqrt((x - x_old)**2 + (y - y_old)** 2)
        
        # 检查收敛条件
        if delta < epsilon:
            break
            
        # 防止过度迭代
        if iterations > 100:
            break
            
    return iter_history, x_history, y_history

# 定义5种不同的学习率
learning_rates = [0.1, 0.3, 0.5, 0.7, 0.9]
colors = ['b', 'g', 'r', 'c', 'm']
labels = [f'学习率 = {lr}' for lr in learning_rates]

# 创建图形
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 10))

# 对每种学习率执行梯度下降并绘图
for i, lr in enumerate(learning_rates):
    iter_history, x_history, y_history = gradient_descent_2d(lr)
    
    # 绘制x值随迭代次数变化曲线
    ax1.plot(iter_history, x_history, color=colors[i], label=labels[i],
             marker='o', markersize=3, linewidth=1.5)
    
    # 绘制y值随迭代次数变化曲线
    ax2.plot(iter_history, y_history, color=colors[i], label=labels[i],
             marker='o', markersize=3, linewidth=1.5)

# 添加参考线（理论最优解）
ax1.axhline(y=1, color='k', linestyle='--', alpha=0.3, label='理论最优x=1')
ax2.axhline(y=-2, color='k', linestyle='--', alpha=0.3, label='理论最优y=-2')

# 设置图表属性
ax1.set_title('x值随迭代次数的收敛曲线', fontsize=14)
ax1.set_xlabel('迭代次数', fontsize=12)
ax1.set_ylabel('x值', fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.legend()

ax2.set_title('y值随迭代次数的收敛曲线', fontsize=14)
ax2.set_xlabel('迭代次数', fontsize=12)
ax2.set_ylabel('y值', fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.show()

# 使用指定参数(学习率0.2)运行并输出结果
print("使用学习率0.2的梯度下降结果：")
iter_history, x_history, y_history = gradient_descent_2d(0.2)
final_x = x_history[-1]
final_y = y_history[-1]
final_f = (final_x - 1)**2 + (final_y + 2)** 2
print(f"最终点: (x={final_x:.6f}, y={final_y:.6f})")
print(f"函数值: f(x,y)={final_f:.6f}")
print(f"迭代次数: {len(iter_history)-1}")
  
```

```python
####2.3节 例5的Python程序 #####
import numpy as np
import matplotlib.pyplot as plt

# 设置中文显示
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False

# 样本数据
X = np.array([1, 2, 3], dtype=np.float32)
y = np.array([1, 1, 0], dtype=np.float32)

# 初始参数
w = 0.5
b = 0.2
learning_rate = 0.1

def sigmoid(z):
    """sigmoid激活函数"""
    return 1 / (1 + np.exp(-z))

def cross_entropy_loss(y_true, y_pred):
    """交叉熵损失函数"""
    # 防止log(0)的情况
    epsilon = 1e-10
    return -np.mean(y_true * np.log(y_pred + epsilon) + 
                   (1 - y_true) * np.log(1 - y_pred + epsilon))

def gradient_descent(X, y, w, b, learning_rate, epochs=100):
    """梯度下降法训练模型"""
    n = len(X)
    w_history = [w]
    b_history = [b]
    loss_history = []
    
    for epoch in range(epochs):
        # 前向传播
        z = w * X + b
        y_pred = sigmoid(z)
        loss = cross_entropy_loss(y, y_pred)
        loss_history.append(loss)
        
        # 计算梯度
        dw = np.mean((y_pred - y) * X)
        db = np.mean(y_pred - y)
        
        # 更新参数
        w -= learning_rate * dw
        b -= learning_rate * db
        
        w_history.append(w)
        b_history.append(b)
        
        # 打印每10轮的结果
        if (epoch + 1) % 10 == 0:
            print(f"梯度下降 - 轮次 {epoch+1}: w={w:.4f}, b={b:.4f}, 损失={loss:.4f}")
    
    return w, b, w_history, b_history, loss_history

def lagrange_method(X, y, w_init, b_init, lambda_reg=0.1, learning_rate=0.1, epochs=100):
    """带L2正则化的拉格朗日乘子法"""
    n = len(X)
    w = w_init
    b = b_init
    w_history = [w]
    b_history = [b]
    loss_history = []
    
    for epoch in range(epochs):
        # 前向传播
        z = w * X + b
        y_pred = sigmoid(z)
        
        # 带正则化的损失函数（拉格朗日函数）
        loss = cross_entropy_loss(y, y_pred) + 0.5 * lambda_reg * w**2
        loss_history.append(loss)
        
        # 计算梯度（包含正则化项的导数）
        dw = np.mean((y_pred - y) * X) + lambda_reg * w
        db = np.mean(y_pred - y)
        
        # 更新参数
        w -= learning_rate * dw
        b -= learning_rate * db
        
        w_history.append(w)
        b_history.append(b)
        
        # 打印每10轮的结果
        if (epoch + 1) % 10 == 0:
            print(f"拉格朗日法 - 轮次 {epoch+1}: w={w:.4f}, b={b:.4f}, 损失={loss:.4f}")
    
    return w, b, w_history, b_history, loss_history

# 使用梯度下降法训练
print("===== 梯度下降法训练 =====")
w_gd, b_gd, w_history_gd, b_history_gd, loss_history_gd = gradient_descent(
    X, y, w, b, learning_rate)

# 使用拉格朗日乘子法（带L2正则化）训练
print("\n===== 拉格朗日乘子法训练 =====")
w_lagrange, b_lagrange, w_history_lagrange, b_history_lagrange, loss_history_lagrange = lagrange_method(
    X, y, w, b, learning_rate=learning_rate)

# 绘制损失曲线对比
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(loss_history_gd, label='梯度下降法')
plt.plot(loss_history_lagrange, label='拉格朗日乘子法')
plt.title('损失函数随迭代次数变化')
plt.xlabel('迭代次数')
plt.ylabel('损失值')
plt.legend()
plt.grid(True, alpha=0.3)

# 绘制参数w的变化曲线
plt.subplot(1, 2, 2)
plt.plot(w_history_gd[1:], label='梯度下降法的w')
plt.plot(w_history_lagrange[1:], label='拉格朗日乘子法的w')
plt.title('参数w随迭代次数变化')
plt.xlabel('迭代次数')
plt.ylabel('w值')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# 最终结果对比
print("\n===== 最终结果对比 =====")
print(f"梯度下降法最终参数: w={w_gd:.4f}, b={b_gd:.4f}")
print(f"拉格朗日乘子法最终参数: w={w_lagrange:.4f}, b={b_lagrange:.4f}")

# 预测结果
print("\n===== 预测结果 =====")
X_test = np.array([1, 2, 3])
y_pred_gd = sigmoid(w_gd * X_test + b_gd)
y_pred_lagrange = sigmoid(w_lagrange * X_test + b_lagrange)

print(f"原始标签: {y}")
print(f"梯度下降法预测概率: {[f'{p:.4f}' for p in y_pred_gd]}")
print(f"拉格朗日法预测概率: {[f'{p:.4f}' for p in y_pred_lagrange]}")
```

## 2.4节   权重范数约束下的神经网络训练 Python程序

```python
######2.4节   权重范数约束下的神经网络训练 Python程序######
import numpy as np
from scipy.optimize import minimize

# 定义sigmoid激活函数
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 定义目标函数：均方误差（拟合点(2,5)）
def objective(params):
    w1, w2, b1, b2 = params
    hidden_output = sigmoid(w1 * 2 + b1)  # 隐藏层输出（输入x=2）
    predicted = w2 * hidden_output + b2   # 输出层预测值
    return (5 - predicted) ** 2  # 均方误差损失

# 等式约束：w1² + w2² = 4
def eq_constraint(params):
    w1, w2, b1, b2 = params
    return w1**2 + w2**2 - 4

# 不等式约束：b1² + b2² ≤ 1 → 转换为 1 - b1² - b2² ≥ 0
def ineq_constraint(params):
    w1, w2, b1, b2 = params
    return 1 - (b1**2 + b2**2)

# 约束条件配置
constraints = [
    {'type': 'eq', 'fun': eq_constraint},   # 等式约束
    {'type': 'ineq', 'fun': ineq_constraint}  # 不等式约束
]

# 初始参数猜测（满足w1²+w2²=4，如w1=2, w2=0；b1=0, b2=0）
initial_guess = [2.0, 0.0, 0.0, 0.0]

# 调用带约束优化求解
result = minimize(objective, initial_guess, method='SLSQP', constraints=constraints)

# 提取并打印结果
print("==== 优化结果 summary ====")
print(result)

print("\n==== 最优参数与预测 ====")
w1_opt, w2_opt, b1_opt, b2_opt = result.x
hidden_out = sigmoid(w1_opt * 2 + b1_opt)
predicted = w2_opt * hidden_out + b2_opt
loss = (5 - predicted) ** 2

print(f"最优权重 w1 = {w1_opt:.4f}, w2 = {w2_opt:.4f}")
print(f"最优偏置 b1 = {b1_opt:.4f}, b2 = {b2_opt:.4f}")
print(f"隐藏层输出：{hidden_out:.4f}")
print(f"最终预测值：{predicted:.4f}")
print(f"最终均方误差：{loss:.4f}")
    
```

```python
import numpy as np

# 定义输入信号和卷积核
x = np.array([5, 5, 5, 10, 10, 10, 5, 5, 5])  # 输入信号
h = np.array([-1, 0, 1])  # 边缘检测核

# 使用 numpy 计算卷积
conv_result = np.convolve(x, h, mode='valid')

# 手动计算卷积（可选，用于理解卷积过程）
def manual_convolve(x, h):
    result = []
    len_x = len(x)
    len_h = len(h)
    for n in range(len_x - len_h + 1):
        sum_val = 0
        for k in range(len_h):
            sum_val += x[n + k] * h[k]
        result.append(sum_val)
    return np.array(result)

manual_result = manual_convolve(x, h)

# 输出结果
print("输入信号 x[n]:", x)
print("卷积核 h[n]:", h)
print("numpy 计算的卷积结果:", conv_result)
print("手动计算的卷积结果:", manual_result)

# 分析卷积结果的意义
print("\n卷积结果分析:")
for i, val in enumerate(conv_result):
    if abs(val) > 0:
        print(f"位置 {i+1} 存在边缘，卷积值为 {val}")
```

```python
# ### Implementing a CNN using the torch.nn module
# 
# #### Configuring CNN layers in PyTorch
# 
#  * **Conv2d:** `torch.nn.Conv2d`
#    * `out_channels`
#    * `kernel_size`
#    * `stride`
#    * `padding`
#    
#    
#  * **MaxPool2d:** `torch.nn.MaxPool2d`
#    * `kernel_size`
#    * `stride`
#    * `padding`
#    
#    
#  * **Dropout** `torch.nn.Dropout`
#    * `p`

# ### Constructing a CNN in PyTorch



model = nn.Sequential()
model.add_module('conv1', nn.Conv2d(in_channels=1, out_channels=32, kernel_size=5, padding=2))
model.add_module('relu1', nn.ReLU())        
model.add_module('pool1', nn.MaxPool2d(kernel_size=2))   
model.add_module('conv2', nn.Conv2d(in_channels=32, out_channels=64, kernel_size=5, padding=2))
model.add_module('relu2', nn.ReLU())        
model.add_module('pool2', nn.MaxPool2d(kernel_size=2))      

x = torch.ones((4, 1, 28, 28))
model(x).shape




model.add_module('flatten', nn.Flatten()) 

x = torch.ones((4, 1, 28, 28))
model(x).shape




model.add_module('fc1', nn.Linear(3136, 1024)) 
model.add_module('relu3', nn.ReLU()) 
model.add_module('dropout', nn.Dropout(p=0.5)) 

model.add_module('fc2', nn.Linear(1024, 10)) 




device = torch.device("cuda:0")
# device = torch.device("cpu")

model = model.to(device) 




loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)
# 优化器的选择

def train(model, num_epochs, train_dl, valid_dl):
    loss_hist_train = [0] * num_epochs
    accuracy_hist_train = [0] * num_epochs
    loss_hist_valid = [0] * num_epochs
    accuracy_hist_valid = [0] * num_epochs
    for epoch in range(num_epochs):
        model.train()
        for x_batch, y_batch in train_dl:
            x_batch = x_batch.to(device) 
            y_batch = y_batch.to(device) 
            pred = model(x_batch)
            loss = loss_fn(pred, y_batch)
            loss.backward()
            optimizer.step()
            optimizer.zero_grad()
            loss_hist_train[epoch] += loss.item()*y_batch.size(0)
            is_correct = (torch.argmax(pred, dim=1) == y_batch).float()
            accuracy_hist_train[epoch] += is_correct.sum().cpu()

        loss_hist_train[epoch] /= len(train_dl.dataset)
        accuracy_hist_train[epoch] /= len(train_dl.dataset)
        
        model.eval()
        with torch.no_grad():
            for x_batch, y_batch in valid_dl:
                x_batch = x_batch.to(device) 
                y_batch = y_batch.to(device) 
                pred = model(x_batch)
                loss = loss_fn(pred, y_batch)
                loss_hist_valid[epoch] += loss.item()*y_batch.size(0) 
                is_correct = (torch.argmax(pred, dim=1) == y_batch).float() 
                accuracy_hist_valid[epoch] += is_correct.sum().cpu()

        loss_hist_valid[epoch] /= len(valid_dl.dataset)
        accuracy_hist_valid[epoch] /= len(valid_dl.dataset)
        
        print(f'Epoch {epoch+1} accuracy: {accuracy_hist_train[epoch]:.4f} val_accuracy: {accuracy_hist_valid[epoch]:.4f}')
    return loss_hist_train, loss_hist_valid, accuracy_hist_train, accuracy_hist_valid

torch.manual_seed(1)
num_epochs = 20 #训练20轮
hist = train(model, num_epochs, train_dl, valid_dl)
```

## 蒙特卡洛算法计算圆周率


```python
import random

def monte_carlo_circle_area(N, r=1):
    K = 0
    square_area = (2*r)**2  # 正方形面积
    for _ in range(N):
        x = random.uniform(-r, r)  # x在[-r, r]随机取值
        y = random.uniform(-r, r)  # y在[-r, r]随机取值
        if x**2 + y**2 <= r**2:  # 判断是否在圆内
            K += 1
    circle_area = square_area * (K / N)
    return circle_area

# 测试：用100万个点估算单位圆面积
N = 1000000
area = monte_carlo_circle_area(N)
print(f"估算圆面积：{area}")  # 约为3.14左右
```

```python
import torch
import torchvision.datasets as dsets
import torchvision.transforms as transforms

batch_size = 64


def load_data():
    """通过调用torchvision中datasets模块来导入MNIST数据集中的训练集和测试集，
    将导入的训练集通过DataLoader加载为train_loader，
    测试集总共有10，000的样本数，分别取5，000作为验证集和测试集，对验证机和测试
    集中样本打乱通过DataLoader加载为validation_loader和test_loader
    """
    train_dataset = dsets.MNIST(root='./data', train=True, transform=transforms.ToTensor(), download=True)
    test_dataset = dsets.MNIST(root='./data', train=False, transform=transforms.ToTensor())

    train_loader = torch.utils.data.DataLoader(dataset=train_dataset, batch_size=batch_size, shuffle=True)

    indices = range(len(test_dataset))
    indices_val = indices[:5000]
    indices_test = indices[5000:]

    sampler_val = torch.utils.data.sampler.SubsetRandomSampler(indices_val)
    sampler_test = torch.utils.data.sampler.SubsetRandomSampler(indices_test)

    validation_loader = torch.utils.data.DataLoader(dataset=test_dataset, batch_size=batch_size, shuffle=False,
                                                    sampler=sampler_val)
    test_loader = torch.utils.data.DataLoader(dataset=test_dataset, batch_size=batch_size, shuffle=False,
                                              sampler=sampler_test)

    return train_loader, validation_loader, test_loader
    
```

```python
# -*- coding: utf-8 -*-
"""通过pytorch中神经网络的模块和函数来构建对MNIST数据集网络的构建、
训练、验证、和测试，整个过程使用了三层的神经元的网络来建立网络；最后
测试集中的正确率有94.36%左右，通过增加网络层数，调整参数，迭代次数，
损失函数等等都能对提高正确率起一定效果

 
"""

import torch
import torch.nn as nn
from torch.autograd import Variable
import torch.optim as optim
import mnist_loader


class Network(nn.Module):
    """这里以[784 30 10]三层神经网络为例
    """
    def __init__(self, sizes):
        super(Network, self).__init__()
        self.sizes = sizes
        self.layer1 = nn.Linear(sizes[0], sizes[1])
        self.layer2 = nn.Linear(sizes[1], sizes[2])

    def forward(self, a):
        a = a.view(-1, self.sizes[0])  # view函数将输入Tensor转换成（64, 784）
        a = self.layer1(a)
        a = self.layer2(a)
        a = torch.log_softmax(a, dim=1)
        return a


def rightness(output, target):
    """输入网络的输出Tensor和目标Tensor，
    比较网络的输出Tensor和目标Tensor中对应相等的结果，
    返回比较结果中匹配正确的个数和整个输出或者目标Tensor
    的长度
    """
    rights = 0
    for index in range(len(target.data)):
        if torch.argmax(output[index]) == target.data[index]:
            rights += 1
    return rights, len(target.data)


def train_model(train_loader, epochs, eta):
    """本函数的功能是训练模型，使用交叉熵的损失函数，和
    随机梯度下降的优化算法，学习率为0.001，动量为0.9
    开始训练循环
    自己改变参数识别效果
    """
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.SGD(net.parameters(), lr=eta, momentum=0.9)

    for epoch in range(epochs):
        train_rights = []  # 记录每次迭代正确的结果和总样本

        for batch_idx, (data, target) in enumerate(train_loader):
            data, target = Variable(data), Variable(target)
            net.train()

            output = net(data)
            loss = criterion(output, target)
            optimizer.zero_grad()  # 清空梯度
            loss.backward()  # 反向传播
            optimizer.step()  # 一步随机梯度下降算法
            right = rightness(output, target)  # 计算一批次准确率中（正确样例数， 总样本数）
            train_rights.append(right)

            if batch_idx % 100 == 0:
                validation_model(validation_loader)

        # 求得整个训练样本中正确的样例总数， 和总样本数，可以通过两者得到训练的正确率
        train_r = (sum([tup[0] for tup in train_rights]), sum([tup[1] for tup in train_rights]))
        print("Epoch {0}: {1}/{2}".format(epoch, train_r[0], train_r[1]))


def validation_model(validation_loader):
    """验证模型
    """
    net.eval()
    val_rights = []

    for data, target in validation_loader:
        data, target = Variable(data), Variable(target)
        output = net(data)
        right = rightness(output, target)
        val_rights.append(right)

    val_r = (sum([tup[0] for tup in val_rights]), sum([tup[1] for tup in val_rights]))
    print("验证集的正确率为{:.2f}%".format(100.0 * val_r[0] / val_r[1]))


def test_model(test_loader):
    """测试模型
    """
    net.eval()
    vals = []
    for data, target in test_loader:
        data, target = Variable(data), Variable(target)
        output = net(data)
        val = rightness(output, target)
        vals.append(val)

    rights = (sum([tup[0] for tup in vals]), sum([tup[1] for tup in vals]))
    print("测试集的正确率为{:.2f}%".format(100.0 * rights[0] / rights[1]))


train_loader, validation_loader, test_loader = mnist_loader.load_data()
net = Network([784, 30, 10])
train_model(train_loader, 20, 0.001)
test_model(test_loader)
```
