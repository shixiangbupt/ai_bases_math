```python
####2.3节 例3的Python程序####
import matplotlib.pyplot as plt
import numpy as np
# 设置中文显示
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False  # 正确显示负号

def gradient_descent(alpha, x0=3.0, epsilon=0.01):
    """
    执行梯度下降算法并返回迭代记录
    参数:
        alpha: 学习率
        x0: 初始x值
        epsilon: 收敛阈值
    返回:
        迭代次数列表和对应的x值列表
    """
    x = x0
    iterations = 0
    x_history = [x]  # 记录x的历史值
    iter_history = [0]  # 记录迭代次数
    
    while True:
        gradient = 2 * x  # 梯度
        x_old = x
        x = x - alpha * gradient  # 参数更新
        
        iterations += 1
        x_history.append(x)
        iter_history.append(iterations)
        
        # 检查收敛条件
        if abs(x - x_old) < epsilon:
            break
            
        # 防止过度迭代（对于可能不收敛的情况）
        if iterations > 100:
            break
            
    return iter_history, x_history

# 定义5种不同的学习率
learning_rates = [0.1, 0.2, 0.4, 0.6, 0.9]
colors = ['b', 'g', 'r', 'c', 'm']
labels = [f'学习率 = {lr}' for lr in learning_rates]

# 创建图形
plt.figure(figsize=(10, 6))

# 对每种学习率执行梯度下降并绘图
for i, lr in enumerate(learning_rates):
    iter_history, x_history = gradient_descent(lr)
    plt.plot(iter_history, x_history, color=colors[i], label=labels[i], 
             marker='o', markersize=4, linewidth=2)

# 添加参考线（理论最优解x=0）
plt.axhline(y=0, color='k', linestyle='--', alpha=0.3, label='理论最优解 x=0')

# 设置图表属性
plt.title('不同学习率下x值随迭代次数的收敛曲线', fontsize=14)
plt.xlabel('迭代次数', fontsize=12)
plt.ylabel('x值', fontsize=12)
plt.grid(True, alpha=0.3)
plt.legend()
plt.ylim(-4, 4)  # 设置y轴范围，便于观察
plt.tight_layout()

# 显示图形
plt.show()
 
```

## 2.3节 例4的Python程序

```python
#2.3节 例4的Python程序   
import matplotlib.pyplot as plt
import numpy as np

# 设置中文显示
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False  # 正确显示负号

def gradient_descent_2d(alpha, x0=3, y0=-1, epsilon=0.01):
    """
    对二元函数f(x,y)=(x-1)²+(y+2)²执行梯度下降
    参数:
        alpha: 学习率
        x0, y0: 初始点坐标
        epsilon: 收敛阈值
    返回:
        迭代次数列表、x值历史、y值历史
    """
    x, y = x0, y0
    iterations = 0
    x_history = [x]
    y_history = [y]
    iter_history = [0]
    
    while True:
        # 计算梯度
        grad_x = 2 * (x - 1)  # x方向梯度
        grad_y = 2 * (y + 2)  # y方向梯度
        
        # 保存当前值
        x_old, y_old = x, y
        
        # 参数更新
        x = x - alpha * grad_x
        y = y - alpha * grad_y
        
        iterations += 1
        x_history.append(x)
        y_history.append(y)
        iter_history.append(iterations)
        
        # 计算变化量（使用欧氏距离）
        delta = np.sqrt((x - x_old)**2 + (y - y_old)** 2)
        
        # 检查收敛条件
        if delta < epsilon:
            break
            
        # 防止过度迭代
        if iterations > 100:
            break
            
    return iter_history, x_history, y_history

# 定义5种不同的学习率
learning_rates = [0.1, 0.3, 0.5, 0.7, 0.9]
colors = ['b', 'g', 'r', 'c', 'm']
labels = [f'学习率 = {lr}' for lr in learning_rates]

# 创建图形
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 10))

# 对每种学习率执行梯度下降并绘图
for i, lr in enumerate(learning_rates):
    iter_history, x_history, y_history = gradient_descent_2d(lr)
    
    # 绘制x值随迭代次数变化曲线
    ax1.plot(iter_history, x_history, color=colors[i], label=labels[i],
             marker='o', markersize=3, linewidth=1.5)
    
    # 绘制y值随迭代次数变化曲线
    ax2.plot(iter_history, y_history, color=colors[i], label=labels[i],
             marker='o', markersize=3, linewidth=1.5)

# 添加参考线（理论最优解）
ax1.axhline(y=1, color='k', linestyle='--', alpha=0.3, label='理论最优x=1')
ax2.axhline(y=-2, color='k', linestyle='--', alpha=0.3, label='理论最优y=-2')

# 设置图表属性
ax1.set_title('x值随迭代次数的收敛曲线', fontsize=14)
ax1.set_xlabel('迭代次数', fontsize=12)
ax1.set_ylabel('x值', fontsize=12)
ax1.grid(True, alpha=0.3)
ax1.legend()

ax2.set_title('y值随迭代次数的收敛曲线', fontsize=14)
ax2.set_xlabel('迭代次数', fontsize=12)
ax2.set_ylabel('y值', fontsize=12)
ax2.grid(True, alpha=0.3)
ax2.legend()

plt.tight_layout()
plt.show()

# 使用指定参数(学习率0.2)运行并输出结果
print("使用学习率0.2的梯度下降结果：")
iter_history, x_history, y_history = gradient_descent_2d(0.2)
final_x = x_history[-1]
final_y = y_history[-1]
final_f = (final_x - 1)**2 + (final_y + 2)** 2
print(f"最终点: (x={final_x:.6f}, y={final_y:.6f})")
print(f"函数值: f(x,y)={final_f:.6f}")
print(f"迭代次数: {len(iter_history)-1}")
  
```

## 2.3节 例5的Python程序

```python
####2.3节 例5的Python程序 #####
import numpy as np
import matplotlib.pyplot as plt

# 设置中文显示
plt.rcParams["font.family"] = ["SimHei", "WenQuanYi Micro Hei", "Heiti TC"]
plt.rcParams["axes.unicode_minus"] = False

# 样本数据
X = np.array([1, 2, 3], dtype=np.float32)
y = np.array([1, 1, 0], dtype=np.float32)

# 初始参数
w = 0.5
b = 0.2
learning_rate = 0.1

def sigmoid(z):
    """sigmoid激活函数"""
    return 1 / (1 + np.exp(-z))

def cross_entropy_loss(y_true, y_pred):
    """交叉熵损失函数"""
    # 防止log(0)的情况
    epsilon = 1e-10
    return -np.mean(y_true * np.log(y_pred + epsilon) + 
                   (1 - y_true) * np.log(1 - y_pred + epsilon))

def gradient_descent(X, y, w, b, learning_rate, epochs=100):
    """梯度下降法训练模型"""
    n = len(X)
    w_history = [w]
    b_history = [b]
    loss_history = []
    
    for epoch in range(epochs):
        # 前向传播
        z = w * X + b
        y_pred = sigmoid(z)
        loss = cross_entropy_loss(y, y_pred)
        loss_history.append(loss)
        
        # 计算梯度
        dw = np.mean((y_pred - y) * X)
        db = np.mean(y_pred - y)
        
        # 更新参数
        w -= learning_rate * dw
        b -= learning_rate * db
        
        w_history.append(w)
        b_history.append(b)
        
        # 打印每10轮的结果
        if (epoch + 1) % 10 == 0:
            print(f"梯度下降 - 轮次 {epoch+1}: w={w:.4f}, b={b:.4f}, 损失={loss:.4f}")
    
    return w, b, w_history, b_history, loss_history

def lagrange_method(X, y, w_init, b_init, lambda_reg=0.1, learning_rate=0.1, epochs=100):
    """带L2正则化的拉格朗日乘子法"""
    n = len(X)
    w = w_init
    b = b_init
    w_history = [w]
    b_history = [b]
    loss_history = []
    
    for epoch in range(epochs):
        # 前向传播
        z = w * X + b
        y_pred = sigmoid(z)
        
        # 带正则化的损失函数（拉格朗日函数）
        loss = cross_entropy_loss(y, y_pred) + 0.5 * lambda_reg * w**2
        loss_history.append(loss)
        
        # 计算梯度（包含正则化项的导数）
        dw = np.mean((y_pred - y) * X) + lambda_reg * w
        db = np.mean(y_pred - y)
        
        # 更新参数
        w -= learning_rate * dw
        b -= learning_rate * db
        
        w_history.append(w)
        b_history.append(b)
        
        # 打印每10轮的结果
        if (epoch + 1) % 10 == 0:
            print(f"拉格朗日法 - 轮次 {epoch+1}: w={w:.4f}, b={b:.4f}, 损失={loss:.4f}")
    
    return w, b, w_history, b_history, loss_history

# 使用梯度下降法训练
print("===== 梯度下降法训练 =====")
w_gd, b_gd, w_history_gd, b_history_gd, loss_history_gd = gradient_descent(
    X, y, w, b, learning_rate)

# 使用拉格朗日乘子法（带L2正则化）训练
print("\n===== 拉格朗日乘子法训练 =====")
w_lagrange, b_lagrange, w_history_lagrange, b_history_lagrange, loss_history_lagrange = lagrange_method(
    X, y, w, b, learning_rate=learning_rate)

# 绘制损失曲线对比
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(loss_history_gd, label='梯度下降法')
plt.plot(loss_history_lagrange, label='拉格朗日乘子法')
plt.title('损失函数随迭代次数变化')
plt.xlabel('迭代次数')
plt.ylabel('损失值')
plt.legend()
plt.grid(True, alpha=0.3)

# 绘制参数w的变化曲线
plt.subplot(1, 2, 2)
plt.plot(w_history_gd[1:], label='梯度下降法的w')
plt.plot(w_history_lagrange[1:], label='拉格朗日乘子法的w')
plt.title('参数w随迭代次数变化')
plt.xlabel('迭代次数')
plt.ylabel('w值')
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# 最终结果对比
print("\n===== 最终结果对比 =====")
print(f"梯度下降法最终参数: w={w_gd:.4f}, b={b_gd:.4f}")
print(f"拉格朗日乘子法最终参数: w={w_lagrange:.4f}, b={b_lagrange:.4f}")

# 预测结果
print("\n===== 预测结果 =====")
X_test = np.array([1, 2, 3])
y_pred_gd = sigmoid(w_gd * X_test + b_gd)
y_pred_lagrange = sigmoid(w_lagrange * X_test + b_lagrange)

print(f"原始标签: {y}")
print(f"梯度下降法预测概率: {[f'{p:.4f}' for p in y_pred_gd]}")
print(f"拉格朗日法预测概率: {[f'{p:.4f}' for p in y_pred_lagrange]}")
```

## 例2.3节 例6的Python程序

```python
######例2.3节 例6的Python程序。#######
import numpy as np

def sigmoid(x):
    """Sigmoid激活函数，用于量化词语间的语义关联程度"""
    return 1 / (1 + np.exp(-x))

# ========== 初始化参数 ==========
v_w = np.array([0.3, 0.4])    # 目标词向量（初始值）
v_c = np.array([0.1, 0.2])    # 上下文词向量（初始值）
v_w1 = np.array([0.5, 0.6])   # 负样本向量（共1个负样本）
learning_rate = 0.05          # 学习率 η


# ========== 计算正/负样本的点积与Sigmoid值 ==========
# 正样本点积与Sigmoid
positive_dot = np.dot(v_w, v_c)
positive_sigmoid = sigmoid(positive_dot)

# 负样本点积与Sigmoid（注意负样本项是 -v_wk · v_c，所以传入 -negative_dot）
negative_dot = np.dot(v_w1, v_c)
negative_sigmoid = sigmoid(-negative_dot)


# ========== 计算损失函数 ==========
loss = -np.log(positive_sigmoid) - np.log(negative_sigmoid)
print(f"负采样损失值 L(θ) = {loss:.4f}")


# ========== 计算梯度（对v_w和v_c的偏导） ==========
# 1. 对目标词向量 v_w 的梯度
gradient_vw = - (positive_sigmoid - 1) * v_c + negative_sigmoid * v_c
print(f"v_w 的梯度 ∇v_w L = {gradient_vw}")

# 2. 对上下文词向量 v_c 的梯度
gradient_vc = - (positive_sigmoid - 1) * v_w + negative_sigmoid * (-v_w1)
print(f"v_c 的梯度 ∇v_c L = {gradient_vc}")


# ========== 参数更新（梯度下降） ==========
v_w_updated = v_w - learning_rate * gradient_vw
v_c_updated = v_c - learning_rate * gradient_vc

print("\n参数更新结果：")
print(f"更新后的目标词向量 v_w = {v_w_updated}")
print(f"更新后的上下文词向量 v_c = {v_c_updated}")
   
```

## 2.3节 例7的Python程序

```python
######$2.3节 例7的Python程序#########
# 初始化参数
p_u = [1.0, 0.5]       # 用户特征向量
q_i = [0.8, 0.3]       # 物品特征向量
learning_rate = 0.02   # 学习率 η
r_ui = 4               # 真实评分

# 1. 计算预测评分（用户向量与物品向量的点积）
pred_score = p_u[0] * q_i[0] + p_u[1] * q_i[1]
print("初始预测评分:", pred_score)

# 2. 计算误差（真实评分 - 预测评分）
error = r_ui - pred_score
print("误差 (r_ui - 预测评分):", error)

# 3. 计算梯度（基于均方误差的梯度推导）
grad_p = [-error * q_i[0], -error * q_i[1]]  # p_u 的梯度
grad_q = [-error * p_u[0], -error * p_u[1]]  # q_i 的梯度
print("p_u 的梯度:", grad_p)
print("q_i 的梯度:", grad_q)

# 4. 梯度下降更新参数
p_u_new = [
    p_u[0] - learning_rate * grad_p[0],
    p_u[1] - learning_rate * grad_p[1]
]
q_i_new = [
    q_i[0] - learning_rate * grad_q[0],
    q_i[1] - learning_rate * grad_q[1]
]

# 输出更新后的参数
print("\n更新后的用户特征向量 p_u:", p_u_new)
print("更新后的物品特征向量 q_i:", q_i_new)

# 5. 计算更新后的预测评分
new_pred_score = p_u_new[0] * q_i_new[0] + p_u_new[1] * q_i_new[1]
print("更新后的预测评分:", new_pred_score)
 
```

## 2.4节   权重范数约束下的神经网络训练 Python程序

```python
######2.4节   权重范数约束下的神经网络训练 Python程序######
import numpy as np
from scipy.optimize import minimize

# 定义sigmoid激活函数
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 定义目标函数：均方误差（拟合点(2,5)）
def objective(params):
    w1, w2, b1, b2 = params
    hidden_output = sigmoid(w1 * 2 + b1)  # 隐藏层输出（输入x=2）
    predicted = w2 * hidden_output + b2   # 输出层预测值
    return (5 - predicted) ** 2  # 均方误差损失

# 等式约束：w1² + w2² = 4
def eq_constraint(params):
    w1, w2, b1, b2 = params
    return w1**2 + w2**2 - 4

# 不等式约束：b1² + b2² ≤ 1 → 转换为 1 - b1² - b2² ≥ 0
def ineq_constraint(params):
    w1, w2, b1, b2 = params
    return 1 - (b1**2 + b2**2)

# 约束条件配置
constraints = [
    {'type': 'eq', 'fun': eq_constraint},   # 等式约束
    {'type': 'ineq', 'fun': ineq_constraint}  # 不等式约束
]

# 初始参数猜测（满足w1²+w2²=4，如w1=2, w2=0；b1=0, b2=0）
initial_guess = [2.0, 0.0, 0.0, 0.0]

# 调用带约束优化求解
result = minimize(objective, initial_guess, method='SLSQP', constraints=constraints)

# 提取并打印结果
print("==== 优化结果 summary ====")
print(result)

print("\n==== 最优参数与预测 ====")
w1_opt, w2_opt, b1_opt, b2_opt = result.x
hidden_out = sigmoid(w1_opt * 2 + b1_opt)
predicted = w2_opt * hidden_out + b2_opt
loss = (5 - predicted) ** 2

print(f"最优权重 w1 = {w1_opt:.4f}, w2 = {w2_opt:.4f}")
print(f"最优偏置 b1 = {b1_opt:.4f}, b2 = {b2_opt:.4f}")
print(f"隐藏层输出：{hidden_out:.4f}")
print(f"最终预测值：{predicted:.4f}")
print(f"最终均方误差：{loss:.4f}")
    
```

## 习题16范例程序：一维卷积特征提取

```python
import numpy as np

# 定义输入信号和卷积核
x = np.array([5, 5, 5, 10, 10, 10, 5, 5, 5])  # 输入信号
h = np.array([-1, 0, 1])  # 边缘检测核

# 使用 numpy 计算卷积
conv_result = np.convolve(x, h, mode='valid')

# 手动计算卷积（可选，用于理解卷积过程）
def manual_convolve(x, h):
    result = []
    len_x = len(x)
    len_h = len(h)
    for n in range(len_x - len_h + 1):
        sum_val = 0
        for k in range(len_h):
            sum_val += x[n + k] * h[k]
        result.append(sum_val)
    return np.array(result)

manual_result = manual_convolve(x, h)

# 输出结果
print("输入信号 x[n]:", x)
print("卷积核 h[n]:", h)
print("numpy 计算的卷积结果:", conv_result)
print("手动计算的卷积结果:", manual_result)

# 分析卷积结果的意义
print("\n卷积结果分析:")
for i, val in enumerate(conv_result):
    if abs(val) > 0:
        print(f"位置 {i+1} 存在边缘，卷积值为 {val}")
```

## 第二章第五节，函数正交性：验证 $\sum_{i=0}^{N-1} \sin(x_i)\cos(x_i) \cdot \Delta x \approx 0$ 

```python
import numpy as np

# 设置参数
N = 8  # 分割数
a = 0
b = 2 * np.pi
delta_x = (b - a) / N

# 生成均匀分布的点
x_values = np.linspace(a, b, N, endpoint=False)

# 计算函数值
f_values = np.sin(x_values) * np.cos(x_values)

# 计算黎曼和
riemann_sum = np.sum(f_values) * delta_x

# 打印结果
print(f"黎曼和结果: {riemann_sum}")
print(f"理论值: 0")

# 验证结果是否接近0
tolerance = 1e-10
if abs(riemann_sum) < tolerance:
    print("验证通过: 黎曼和与理论值一致。")
else:
    print("验证失败: 黎曼和与理论值存在显著差异。")
```

## 蒙特卡洛算法计算圆周率


```python
import random

def monte_carlo_circle_area(N, r=1):
    K = 0
    square_area = (2*r)**2  # 正方形面积
    for _ in range(N):
        x = random.uniform(-r, r)  # x在[-r, r]随机取值
        y = random.uniform(-r, r)  # y在[-r, r]随机取值
        if x**2 + y**2 <= r**2:  # 判断是否在圆内
            K += 1
    circle_area = square_area * (K / N)
    return circle_area

# 测试：用100万个点估算单位圆面积
N = 1000000
area = monte_carlo_circle_area(N)
print(f"估算圆面积：{area}")  # 约为3.14左右
```
